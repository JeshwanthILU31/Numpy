```python
import numpy as np

s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
#row_sum 
b = np.sum(s,axis=1) 
for i in b:
    if i != 45:
        print("sudoku is not valid")
        break 
else:
    print("for rows it is valid")
```

```python
c = np.sum(s,axis = 0) 
for j in c:
    if j != 45:
        print("sudoku is not valid")
        break 
else:
    print("for cols it is valid")
```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 91, 85]    # Student 5
])
```

```python
#get the shape of matrix.
data.shape
```

```python
#find the average age of students.
np.mean(data[:,0])
```

```python
#exact maths marks of all students.
data[:,1]
```

```python
#find the highest science marks
int(np.max(data[:,2]))
```

```python
#get the details of the student who scored more than 90 marks in Math
data[data[:,1] > 90]
```

```python
#Increase Math marks of all students by 5 
data[:,1]+=5
data
```

```python
#find how many students are younger than 19 
len(data[data[:,0] < 19])
```

```python
#calculate the average marks in each subject (column-wise mean).
```

```python
np.mean(data[:,1:],axis = 0)
```

```python
#get data of the students who scored at least 80 in both subjects
data[(data[:,1] >= 80) & (data[:,2] >= 80)]
```

```python
#Replace all Science marks < 75 with 0

data[:,2][data[:,2] < 75] = 0
data
```
